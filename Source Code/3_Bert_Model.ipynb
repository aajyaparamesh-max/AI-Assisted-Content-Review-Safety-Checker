#**Function to Save results to Excel**

```python
import os
import pandas as pd

# ==========================================================
# RESULT FILE LOCATION
# ==========================================================

RESULT_FOLDER = f"{PROJECT_DIR}/data"
RESULT_FILE = f"{PROJECT_DIR}/data/ModelComparison.csv"  #os.path.join(RESULT_FOLDER, "ModelComparison.csv")

os.makedirs(RESULT_FOLDER, exist_ok=True)

# ==========================================================
# FUNCTION TO SAVE / UPDATE RESULTS
# ==========================================================

def update_results_to_csv(metrics):

    # ------------------------------------------------------
    # First execution -> Create Excel automatically
    # ------------------------------------------------------

    if not os.path.exists(RESULT_FILE):

        model_results = pd.DataFrame([metrics])

    else:

        model_results = pd.read_csv(RESULT_FILE)

        # Remove previous result of the same model
        if "Model" in model_results.columns:
            model_results = model_results[
                model_results["Model"] != metrics["Model"]
            ]

        # Remove Rank column (it will be recreated)
        if "Rank" in model_results.columns:
            model_results = model_results.drop(columns=["Rank"])

        # Append new result
        model_results = pd.concat(
            [model_results, pd.DataFrame([metrics])],
            ignore_index=True
        )

    # ------------------------------------------------------
    # Sort by performance
    # ------------------------------------------------------

    model_results = model_results.sort_values(
        by=["Micro F1", "Macro F1", "Accuracy"],
        ascending=[False, False, False]
    ).reset_index(drop=True)

    # ------------------------------------------------------
    # Create Rank
    # ------------------------------------------------------

    model_results.insert(0, "Rank", range(1, len(model_results) + 1))

    # Move Rank column to first position
    cols = ["Rank"] + [
        c for c in model_results.columns
        if c != "Rank"
    ]

    model_results = model_results[cols]

    # ------------------------------------------------------
    # Round numeric values
    # ------------------------------------------------------

    numeric_cols = model_results.select_dtypes(
        include="number"
    ).columns

    model_results[numeric_cols] = (
        model_results[numeric_cols]
        .round(4)
    )

    # ------------------------------------------------------
    # Save Excel
    # ------------------------------------------------------

    model_results.to_csv(RESULT_FILE, index=False)

    print("=" * 60)
    print(f"{metrics['Model']} results saved successfully.")
    print(f"Results File : {RESULT_FILE}")
    print("=" * 60)

    return model_results
```

#**Evaluation Function**

```python
results = []

def evaluate_model(
    model_name,
    y_true,
    y_pred,
    y_prob=None
):

    metrics = {

        "Model":model_name,

        "Accuracy":
        accuracy_score(y_true,y_pred),

        "Micro Precision":
        precision_score(
            y_true,
            y_pred,
            average='micro',
            zero_division=0
        ),

        "Micro Recall":
        recall_score(
            y_true,
            y_pred,
            average='micro',
            zero_division=0
        ),

        "Micro F1":
        f1_score(
            y_true,
            y_pred,
            average='micro',
            zero_division=0
        ),

        "Macro F1":
        f1_score(
            y_true,
            y_pred,
            average='macro',
            zero_division=0
        ),

        "Weighted F1":
        f1_score(
            y_true,
            y_pred,
            average='weighted',
            zero_division=0
        ),

        "Hamming Loss":
        hamming_loss(
            y_true,
            y_pred
        ),

        "Jaccard Score":
        jaccard_score(
            y_true,
            y_pred,
            average='samples',
            zero_division=0
        )
    }

    if y_prob is not None:
        metrics["ROC-AUC"] = roc_auc_score(
            y_true,
            y_prob,
            average="micro"
        )
    else:
        metrics["ROC-AUC"] = np.nan

    results.append(metrics)

    update_results_to_csv(metrics)

    print("="*70)
    print(model_name)
    print("="*70)

    for k,v in metrics.items():

        if k!="Model":

            print(f"{k:<20}: {v:.4f}")
```

```python
# Step 6: Tokenize Text
MAX_LENGTH = 128 #64

train_encodings = tokenizer(
    X_train,
    truncation=True,
    padding=True,
    max_length=MAX_LENGTH
)

test_encodings = tokenizer(
    X_test,
    truncation=True,
    padding=True,
    max_length=MAX_LENGTH
)
```

```python
# Step 7: Create Dataset Class
class ToxicDataset(Dataset):

    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):

        item = {
            key: torch.tensor(value[idx])
            for key, value in self.encodings.items()
        }

        item["labels"] = torch.tensor(
            self.labels[idx],
            dtype=torch.float
        )

        return item
```

```python
# Step 8: Create DataLoaders
train_dataset = ToxicDataset(
    train_encodings,
    y_train
)

test_dataset = ToxicDataset(
    test_encodings,
    y_test
)

train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

# learning rate, batch size other hyper parameters to be included

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)
```

```python
# ============================================================
# STEP 9 : LOAD BERT MODEL (OPTIMIZED)
# ============================================================


# ============================================================
# BERT CONFIGURATION
# ============================================================

config = BertConfig.from_pretrained(
    "bert-base-uncased",
    num_labels=6,
    problem_type="multi_label_classification",
    hidden_dropout_prob=0.3,
    attention_probs_dropout_prob=0.3
)

bertModel = BertForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    config=config
)

bertModel.to(device)

print("Model Loaded Successfully")
print(f"Running on : {device}")

# ============================================================
# FREEZE LOWER BERT LAYERS
# ============================================================

freeze_layers = 8

for layer in bertModel.bert.encoder.layer[:freeze_layers]:
    for param in layer.parameters():
        param.requires_grad = False

print(f"First {freeze_layers} encoder layers frozen.")

# ============================================================
# TRAINABLE PARAMETERS
# ============================================================

trainable_params = sum(
    p.numel()
    for p in bertModel.parameters()
    if p.requires_grad
)

total_params = sum(
    p.numel()
    for p in bertModel.parameters()
)

print(f"Trainable Parameters : {trainable_params:,}")
print(f"Total Parameters     : {total_params:,}")

# ============================================================
# HYPERPARAMETERS
# ============================================================

EPOCHS = 3

LEARNING_RATE = 2e-5

WEIGHT_DECAY = 0.01

MAX_GRAD_NORM = 1.0

EARLY_STOPPING_PATIENCE = 2

# ============================================================
# OPTIMIZER
# ============================================================

optimizer = AdamW(
    bertModel.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

# ============================================================
# LEARNING RATE SCHEDULER
# ============================================================

total_steps = len(train_loader) * EPOCHS

warmup_steps = int(0.1 * total_steps)

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_steps
)

print(f"Training Batches : {len(train_loader)}")
print(f"Validation Batches : {len(test_loader)}")
print(f"Total Steps : {total_steps}")
print(f"Warmup Steps : {warmup_steps}")

# ============================================================
# MIXED PRECISION (FP16)
# ============================================================

scaler = GradScaler()

print("Mixed Precision Enabled")

# ============================================================
# EARLY STOPPING VARIABLES
# ============================================================

val_loss = float("inf")

epochs_without_improvement = 0

history = {
    "train_loss": [],
    "val_loss": []
}

# ============================================================
# STORE MODEL WEIGHTS
# ============================================================

model_weights = copy.deepcopy(
    bertModel.state_dict()
)

# ============================================================
# GPU INFORMATION
# ============================================================

if torch.cuda.is_available():

    print("=" * 60)
    print("GPU INFORMATION")
    print("=" * 60)

    print("GPU :", torch.cuda.get_device_name(0))

    print(
        "Allocated Memory :",
        round(torch.cuda.memory_allocated(0) / 1024**3, 2),
        "GB"
    )

    print(
        "Cached Memory :",
        round(torch.cuda.memory_reserved(0) / 1024**3, 2),
        "GB"
    )

print("=" * 60)
print("BERT INITIALIZATION COMPLETED")
print("=" * 60)
```

```python
# ============================================================
# TRAINING LOOP
# ============================================================

print("=" * 70)
print("Starting BERT Training...")
print("=" * 70)

start_time = time.time()

for epoch in range(EPOCHS):

    print(f"\nEpoch {epoch+1}/{EPOCHS}")
    print("-" * 70)

    ###########################################################
    # TRAINING
    ###########################################################

    bertModel.train()

    train_loss = 0

    progress_bar = tqdm(
        train_loader,
        desc=f"Training Epoch {epoch+1}",
        leave=True
    )

    for batch in progress_bar:

        input_ids = batch["input_ids"].to(device, non_blocking=True)
        attention_mask = batch["attention_mask"].to(device, non_blocking=True)
        labels = batch["labels"].float().to(device, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)

        # Mixed Precision Forward Pass
        with autocast():

            outputs = bertModel(
                input_ids=input_ids,
                attention_mask=attention_mask,
                labels=labels
            )

            loss = outputs.loss

        train_loss += loss.item()

        # Backpropagation
        scaler.scale(loss).backward()

        # Gradient Clipping
        scaler.unscale_(optimizer)

        torch.nn.utils.clip_grad_norm_(
            bertModel.parameters(),
            MAX_GRAD_NORM
        )

        scaler.step(optimizer)

        scaler.update()

        scheduler.step()

        progress_bar.set_postfix(
            Loss=f"{loss.item():.4f}",
            LR=f"{scheduler.get_last_lr()[0]:.2e}"
        )

    ###########################################################
    # TRAINING LOSS
    ###########################################################

    avg_train_loss = train_loss / len(train_loader)

    ###########################################################
    # VALIDATION
    ###########################################################

    bertModel.eval()

    validation_loss = 0

    with torch.no_grad():

        for batch in tqdm(
            test_loader,
            desc="Validation",
            leave=False
        ):

            input_ids = batch["input_ids"].to(device, non_blocking=True)

            attention_mask = batch["attention_mask"].to(device, non_blocking=True)

            labels = batch["labels"].float().to(device, non_blocking=True)

            with autocast():

                outputs = bertModel(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    labels=labels
                )

                loss = outputs.loss

            validation_loss += loss.item()

    avg_val_loss = validation_loss / len(test_loader)

    ###########################################################
    # STORE HISTORY
    ###########################################################

    history["train_loss"].append(avg_train_loss)
    history["val_loss"].append(avg_val_loss)

    ###########################################################
    # PRINT METRICS
    ###########################################################

    print(f"\nTraining Loss   : {avg_train_loss:.4f}")
    print(f"Validation Loss : {avg_val_loss:.4f}")

    ###########################################################
    # SAVE MODEL
    ###########################################################

    if avg_val_loss < val_loss:

        print("Validation improved.")
        print("Saving Model...")

        val_loss = avg_val_loss

        epochs_without_improvement = 0

        model_weights = copy.deepcopy(
            bertModel.state_dict()
        )

        torch.save(model_weights,f"{PROJECT_DIR}/models/bert_model.pt")

        bertModel.save_pretrained(f"{PROJECT_DIR}/models/bert")

        tokenizer.save_pretrained(f"{PROJECT_DIR}/models/bert")

    else:

        epochs_without_improvement += 1

        print(
            f"No Improvement "
            f"({epochs_without_improvement}/{EARLY_STOPPING_PATIENCE})"
        )

    ###########################################################
    # EARLY STOPPING
    ###########################################################

    if epochs_without_improvement >= EARLY_STOPPING_PATIENCE:

        print("\nEarly Stopping Triggered.")

        break

###############################################################
# TRAINING COMPLETE
###############################################################

training_time = time.time() - start_time

print("=" * 70)
print("Training Completed")
print("=" * 70)

print(f"Validation Loss : {val_loss:.4f}")

print(
    f"Training Time : "
    f"{training_time/60:.2f} minutes"
)

###############################################################
# LOAD MODEL
###############################################################

bertModel.load_state_dict(model_weights)

bertModel.eval()

print("\BERT Model Loaded Successfully.")
```

```python
# ============================================================
# LOAD MODEL FOR EVALUATION
# ============================================================

print("="*70)
print("Loading BERT Model")
print("="*70)

bertModel = BertForSequenceClassification.from_pretrained(f"{PROJECT_DIR}/models/bert")

bertModel.to(device)
bertModel.eval()

print("Model loaded successfully.")

# ============================================================
# PREDICTION
# ============================================================

all_labels = []
all_predictions = []
all_probabilities = []

prediction_start = time.time()

with torch.no_grad():

    progress_bar = tqdm(
        test_loader,
        desc="Predicting"
    )

    for batch in progress_bar:

        input_ids = batch["input_ids"].to(device, non_blocking=True)

        attention_mask = batch["attention_mask"].to(
            device,
            non_blocking=True
        )

        labels = batch["labels"].to(device)

        with autocast():

            outputs = bertModel(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            logits = outputs.logits

        probabilities = torch.sigmoid(logits)

        predictions = (probabilities >= 0.50).int()

        all_probabilities.extend(
            probabilities.cpu().numpy()
        )

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_labels.extend(
            labels.cpu().numpy()
        )

prediction_time = time.time() - prediction_start

print(f"\nPrediction Time : {prediction_time:.2f} seconds")

# ============================================================
# CONVERT TO NUMPY
# ============================================================

y_true = np.array(all_labels)

y_pred = np.array(all_predictions)

y_prob = np.array(all_probabilities)

print("\nPrediction Shapes")

print("y_true :", y_true.shape)

print("y_pred :", y_pred.shape)

print("y_prob :", y_prob.shape)

# ============================================================
# MODEL EVALUATION
# ============================================================

print("="*70)
print("BERT MODEL PERFORMANCE")
print("="*70)

evaluate_model(

    model_name="BERT",

    y_true=y_true,

    y_pred=y_pred,

    y_prob=y_prob

)

```
