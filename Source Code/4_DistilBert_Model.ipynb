#**Function to Save results to CSV**

```python
import os
import pandas as pd

# ==========================================================
# RESULT FILE LOCATION
# ==========================================================

RESULT_FOLDER = f"{PROJECT_DIR}/data"
RESULT_FILE = f"{PROJECT_DIR}/data/ModelComparison.xlsx"  #os.path.join(RESULT_FOLDER, "ModelComparison.xlsx")

os.makedirs(RESULT_FOLDER, exist_ok=True)

# ==========================================================
# FUNCTION TO SAVE / UPDATE RESULTS
# ==========================================================

def update_results_to_csv(metrics):

    # ------------------------------------------------------
    # First execution -> Create Excel automatically
    # ------------------------------------------------------

    if not os.path.exists(RESULT_FILE):

        model_results = pd.DataFrame([metrics])

    else:

        model_results = pd.read_csv(RESULT_FILE)

        # Remove existing record for same model
        model_results = model_results[
            model_results["Model"] != metrics["Model"]
        ]

        # Append new row
        model_results = pd.concat(
            [model_results, pd.DataFrame([metrics])],
            ignore_index=True
        )

    # ------------------------------------------------------
    # Sort by performance
    # ------------------------------------------------------

    model_results = model_results.sort_values(
        by=["Micro F1", "Macro F1", "Accuracy"],
        ascending=[False, False, False]
    ).reset_index(drop=True)

    # ------------------------------------------------------
    # Update Rank
    # ------------------------------------------------------

    model_results["Rank"] = range(1, len(model_results) + 1)

    # Move Rank column to first position
    cols = ["Rank"] + [
        c for c in model_results.columns
        if c != "Rank"
    ]

    model_results = model_results[cols]

    # ------------------------------------------------------
    # Round numeric values
    # ------------------------------------------------------

    numeric_cols = model_results.select_dtypes(
        include="number"
    ).columns

    model_results[numeric_cols] = (
        model_results[numeric_cols]
        .round(4)
    )

    # ------------------------------------------------------
    # Save Excel
    # ------------------------------------------------------

    model_results.to_csv(RESULT_FILE,index=False)

    print("=" * 60)
    print(f"{metrics['Model']} results saved successfully.")
    print(f"Results File : {RESULT_FILE}")
    print("=" * 60)

    return model_results
```

```python
import os
import pandas as pd

# ==========================================================
# RESULT FILE LOCATION
# ==========================================================

RESULT_FOLDER = f"{PROJECT_DIR}/data"
RESULT_FILE = f"{PROJECT_DIR}/data/ModelComparison.csv"  #os.path.join(RESULT_FOLDER, "ModelComparison.csv")

os.makedirs(RESULT_FOLDER, exist_ok=True)

# ==========================================================
# FUNCTION TO SAVE / UPDATE RESULTS
# ==========================================================

def update_results_to_csv(metrics):

    # ------------------------------------------------------
    # First execution -> Create Excel automatically
    # ------------------------------------------------------

    if not os.path.exists(RESULT_FILE):

        model_results = pd.DataFrame([metrics])

    else:

        model_results = pd.read_csv(RESULT_FILE)

        # Remove previous result of the same model
        if "Model" in model_results.columns:
            model_results = model_results[
                model_results["Model"] != metrics["Model"]
            ]

        # Remove Rank column (it will be recreated)
        if "Rank" in model_results.columns:
            model_results = model_results.drop(columns=["Rank"])

        # Append new result
        model_results = pd.concat(
            [model_results, pd.DataFrame([metrics])],
            ignore_index=True
        )

    # ------------------------------------------------------
    # Sort by performance
    # ------------------------------------------------------

    model_results = model_results.sort_values(
        by=["Micro F1", "Macro F1", "Accuracy"],
        ascending=[False, False, False]
    ).reset_index(drop=True)

    # ------------------------------------------------------
    # Create Rank
    # ------------------------------------------------------

    model_results.insert(0, "Rank", range(1, len(model_results) + 1))

    # Move Rank column to first position
    cols = ["Rank"] + [
        c for c in model_results.columns
        if c != "Rank"
    ]

    model_results = model_results[cols]

    # ------------------------------------------------------
    # Round numeric values
    # ------------------------------------------------------

    numeric_cols = model_results.select_dtypes(
        include="number"
    ).columns

    model_results[numeric_cols] = (
        model_results[numeric_cols]
        .round(4)
    )

    # ------------------------------------------------------
    # Save Excel
    # ------------------------------------------------------

    model_results.to_csv(RESULT_FILE, index=False)

    print("=" * 60)
    print(f"{metrics['Model']} results saved successfully.")
    print(f"Results File : {RESULT_FILE}")
    print("=" * 60)

    return model_results
```

#**Evaluation Function**

```python
results = []

def evaluate_model(
    model_name,
    y_true,
    y_pred,
    y_prob=None
):

    metrics = {

        "Model":model_name,

        "Accuracy":
        accuracy_score(y_true,y_pred),

        "Micro Precision":
        precision_score(
            y_true,
            y_pred,
            average='micro',
            zero_division=0
        ),

        "Micro Recall":
        recall_score(
            y_true,
            y_pred,
            average='micro',
            zero_division=0
        ),

        "Micro F1":
        f1_score(
            y_true,
            y_pred,
            average='micro',
            zero_division=0
        ),

        "Macro F1":
        f1_score(
            y_true,
            y_pred,
            average='macro',
            zero_division=0
        ),

        "Weighted F1":
        f1_score(
            y_true,
            y_pred,
            average='weighted',
            zero_division=0
        ),

        "Hamming Loss":
        hamming_loss(
            y_true,
            y_pred
        ),

        "Jaccard Score":
        jaccard_score(
            y_true,
            y_pred,
            average='samples',
            zero_division=0
        )
    }

    if y_prob is not None:
        metrics["ROC-AUC"] = roc_auc_score(
            y_true,
            y_prob,
            average="micro"
        )
    else:
        metrics["ROC-AUC"] = np.nan

    results.append(metrics)

    update_results_to_csv(metrics)

    print("="*70)
    print(model_name)
    print("="*70)

    for k,v in metrics.items():

        if k!="Model":

            print(f"{k:<20}: {v:.4f}")
```

```python
# Import Libraries

import torch
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from tqdm.auto import tqdm

from transformers import (
    DistilBertConfig,
    DistilBertTokenizerFast,
    DistilBertForSequenceClassification,
    get_linear_schedule_with_warmup
)

from torch.optim import AdamW
from torch.cuda.amp import autocast, GradScaler

import copy
import time
```

```python
# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print(device)
```

```python
# Load DistilBERT Tokenizer
tokenizer = DistilBertTokenizerFast.from_pretrained("distilbert-base-uncased", force_download=True)

tokenizer.save_pretrained(f"{PROJECT_DIR}/models/bert")
```

```python
# Tokenize Text
MAX_LENGTH = 128

train_encodings = tokenizer(
    X_train,
    truncation=True,
    padding="longest",
    max_length=MAX_LENGTH
)

test_encodings = tokenizer(
    X_test,
    truncation=True,
    padding="longest",
    max_length=MAX_LENGTH
)
```

```python
# Dataset Class
# You can reuse the same class you created for BERT:

class ToxicDataset(Dataset):

    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):

        item = {
            key: torch.tensor(val[idx])
            for key, val in self.encodings.items()
        }

        item["labels"] = torch.tensor(
            self.labels[idx],
            dtype=torch.float
        )

        return item
```

```python
# Create DataLoaders
train_dataset = ToxicDataset(train_encodings, y_train)
test_dataset = ToxicDataset(test_encodings, y_test)

train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)
```

```python
# ============================================================
# LOAD TOKENIZER
# ============================================================

distilTokenizer = DistilBertTokenizerFast.from_pretrained(
    "distilbert-base-uncased"
)

# ============================================================
# DISTILBERT CONFIGURATION
# ============================================================

config = DistilBertConfig.from_pretrained(
    "distilbert-base-uncased",
    num_labels=6,
    problem_type="multi_label_classification",
    dropout=0.3,
    attention_dropout=0.3
)

distilBertModel = DistilBertForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    config=config
)

distilBertModel.to(device)

print("="*60)
print("DistilBERT Loaded Successfully")
print("="*60)

# ============================================================
# FREEZE LOWER TRANSFORMER LAYERS
# ============================================================

freeze_layers = 3      # DistilBERT has 6 transformer layers

for layer in distilBertModel.distilbert.transformer.layer[:freeze_layers]:

    for param in layer.parameters():

        param.requires_grad = False

print(f"Frozen Transformer Layers : {freeze_layers}")

# ============================================================
# TRAINABLE PARAMETERS
# ============================================================

trainable_params = sum(

    p.numel()

    for p in distilBertModel.parameters()

    if p.requires_grad

)

total_params = sum(

    p.numel()

    for p in distilBertModel.parameters()

)

print(f"Trainable Parameters : {trainable_params:,}")

print(f"Total Parameters     : {total_params:,}")

# ============================================================
# HYPERPARAMETERS
# ============================================================

EPOCHS = 3

LEARNING_RATE = 2e-5

WEIGHT_DECAY = 0.01

MAX_GRAD_NORM = 1.0

EARLY_STOPPING_PATIENCE = 2

# ============================================================
# OPTIMIZER
# ============================================================

optimizer = AdamW(

    distilBertModel.parameters(),

    lr=LEARNING_RATE,

    weight_decay=WEIGHT_DECAY

)

# ============================================================
# SCHEDULER
# ============================================================

total_steps = len(train_loader) * EPOCHS

warmup_steps = int(0.10 * total_steps)

scheduler = get_linear_schedule_with_warmup(

    optimizer,

    num_warmup_steps=warmup_steps,

    num_training_steps=total_steps

)

print(f"Training Batches   : {len(train_loader)}")

print(f"Validation Batches : {len(test_loader)}")

print(f"Total Steps        : {total_steps}")

print(f"Warmup Steps       : {warmup_steps}")

# ============================================================
# MIXED PRECISION
# ============================================================

scaler = GradScaler()

print("Mixed Precision Enabled")

# ============================================================
# EARLY STOPPING
# ============================================================

val_loss = float("inf")

epochs_without_improvement = 0

history = {

    "train_loss": [],

    "val_loss": []

}

model_weights = copy.deepcopy(

    distilBertModel.state_dict()

)

# ============================================================
# GPU INFORMATION
# ============================================================

if torch.cuda.is_available():

    print("="*60)

    print("GPU INFORMATION")

    print("="*60)

    print("GPU :", torch.cuda.get_device_name(0))

    print(
        "Allocated Memory :",
        round(torch.cuda.memory_allocated()/1024**3,2),
        "GB"
    )

    print(
        "Reserved Memory :",
        round(torch.cuda.memory_reserved()/1024**3,2),
        "GB"
    )

print("="*60)
print("DISTILBERT INITIALIZATION COMPLETED")
print("="*60)
```

```python
# ============================================================
# DISTILBERT TRAINING LOOP
# ============================================================

print("=" * 70)
print("Starting DistilBERT Training...")
print("=" * 70)

start_time = time.time()

for epoch in range(EPOCHS):

    print(f"\nEpoch {epoch+1}/{EPOCHS}")
    print("-" * 70)

    ###########################################################
    # TRAINING
    ###########################################################

    distilBertModel.train()

    train_loss = 0

    progress_bar = tqdm(
        train_loader,
        desc=f"Training Epoch {epoch+1}",
        leave=True
    )

    for batch in progress_bar:

        input_ids = batch["input_ids"].to(device, non_blocking=True)

        attention_mask = batch["attention_mask"].to(
            device,
            non_blocking=True
        )

        labels = batch["labels"].float().to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(set_to_none=True)

        #######################################################
        # FORWARD PASS (Mixed Precision)
        #######################################################

        with autocast():

            outputs = distilBertModel(

                input_ids=input_ids,

                attention_mask=attention_mask,

                labels=labels

            )

            loss = outputs.loss

        train_loss += loss.item()

        #######################################################
        # BACKWARD PASS
        #######################################################

        scaler.scale(loss).backward()

        scaler.unscale_(optimizer)

        torch.nn.utils.clip_grad_norm_(

            distilBertModel.parameters(),

            MAX_GRAD_NORM

        )

        scaler.step(optimizer)

        scaler.update()

        scheduler.step()

        progress_bar.set_postfix(

            Loss=f"{loss.item():.4f}",

            LR=f"{scheduler.get_last_lr()[0]:.2e}"

        )

    ###########################################################
    # TRAINING LOSS
    ###########################################################

    avg_train_loss = train_loss / len(train_loader)

    ###########################################################
    # VALIDATION
    ###########################################################

    distilBertModel.eval()

    validation_loss = 0

    with torch.no_grad():

        validation_bar = tqdm(

            test_loader,

            desc="Validation",

            leave=False

        )

        for batch in validation_bar:

            input_ids = batch["input_ids"].to(
                device,
                non_blocking=True
            )

            attention_mask = batch["attention_mask"].to(
                device,
                non_blocking=True
            )

            labels = batch["labels"].float().to(
                device,
                non_blocking=True
            )

            with autocast():

                outputs = distilBertModel(

                    input_ids=input_ids,

                    attention_mask=attention_mask,

                    labels=labels

                )

                loss = outputs.loss

            validation_loss += loss.item()

    ###########################################################
    # AVERAGE LOSSES
    ###########################################################

    avg_val_loss = validation_loss / len(test_loader)

    history["train_loss"].append(avg_train_loss)

    history["val_loss"].append(avg_val_loss)

    ###########################################################
    # DISPLAY RESULTS
    ###########################################################

    print()

    print(f"Training Loss   : {avg_train_loss:.4f}")

    print(f"Validation Loss : {avg_val_loss:.4f}")

    ###########################################################
    # SAVE MODEL
    ###########################################################

    if avg_val_loss < val_loss:

        print("\nValidation Improved")

        val_loss = avg_val_loss

        epochs_without_improvement = 0

        model_weights = copy.deepcopy(

            distilBertModel.state_dict()

        )

        #######################################################
        # SAVE MODEL WEIGHTS
        #######################################################

        torch.save(

            model_weights,f"{PROJECT_DIR}/models/distilbert_model.pt")

        #######################################################
        # SAVE COMPLETE MODEL
        #######################################################

        distilBertModel.save_pretrained(f"{PROJECT_DIR}/models/distilbert")

        #######################################################
        # SAVE TOKENIZER
        #######################################################

        distilTokenizer.save_pretrained(f"{PROJECT_DIR}/models/distilbert")

        print("DistilBERT Model Saved")

    else:

        epochs_without_improvement += 1

        print(

            f"\nNo Improvement "

            f"({epochs_without_improvement}/{EARLY_STOPPING_PATIENCE})"

        )

    ###########################################################
    # EARLY STOPPING
    ###########################################################

    if epochs_without_improvement >= EARLY_STOPPING_PATIENCE:

        print("\nEarly Stopping Triggered")

        break

###############################################################
# TRAINING FINISHED
###############################################################

training_time = time.time() - start_time

print()

print("=" * 70)

print("DistilBERT Training Completed")

print("=" * 70)

print(f"Validation Loss : {val_loss:.4f}")

print(

    f"Training Time : "

    f"{training_time/60:.2f} minutes"

)

###############################################################
# LOAD MODEL
###############################################################

distilBertModel.load_state_dict(

    model_weights

)

distilBertModel.eval()

print("\nDistilBERT Model Loaded Successfully")
```
