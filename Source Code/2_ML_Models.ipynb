#**TF-IDF Feature Extraction**

It gives more importance to informative words while reducing the influence of common words.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Build TF-IDF Features

tfidf = TfidfVectorizer(
    # max_features=10000,
    max_features=3000,
    min_df=5,
    max_df=0.90,
    ngram_range=(1,2)
)

X_tfidf = tfidf.fit_transform(df['clean_text'])
```

```python
# Function to Save file
import joblib

def save_joblib(obj, file_path):
    # Delete existing file
    if os.path.exists(file_path):
        os.remove(file_path)
        print(f"Deleted: {file_path}")

    # Save new file
    joblib.dump(obj, file_path)
    print(f"Saved: {file_path}")
```

#**Train/Test Split**

***Multi-label Stratified Split***

```python
# # Multi-label Classification
# not using Binary Classification as there are multiple lables(Toxic, Severe Toxic, Threat, Offensive, Insult, Hate) not just Safe Vs Unsafe

msss = MultilabelStratifiedShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

# Convert COO → CSR
X = X.tocsr()

for train_idx, test_idx in msss.split(X_final, y):

    X_train = X[train_idx]
    X_test = X[test_idx]

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]
```

#**Function to Save results to Excel**

```python
import os
import pandas as pd

# ==========================================================
# RESULT FILE LOCATION
# ==========================================================

RESULT_FOLDER = f"{PROJECT_DIR}/data"
RESULT_FILE = f"{PROJECT_DIR}/data/ModelComparison.csv"  #os.path.join(RESULT_FOLDER, "ModelComparison.csv")

os.makedirs(RESULT_FOLDER, exist_ok=True)

# ==========================================================
# FUNCTION TO SAVE / UPDATE RESULTS
# ==========================================================

def update_results_to_csv(metrics):

    # ------------------------------------------------------
    # First execution -> Create Excel automatically
    # ------------------------------------------------------

    if not os.path.exists(RESULT_FILE):

        model_results = pd.DataFrame([metrics])

    else:

        model_results = pd.read_csv(RESULT_FILE)

        # Remove previous result of the same model
        if "Model" in model_results.columns:
            model_results = model_results[
                model_results["Model"] != metrics["Model"]
            ]

        # Remove Rank column (it will be recreated)
        if "Rank" in model_results.columns:
            model_results = model_results.drop(columns=["Rank"])

        # Append new result
        model_results = pd.concat(
            [model_results, pd.DataFrame([metrics])],
            ignore_index=True
        )

    # ------------------------------------------------------
    # Sort by performance
    # ------------------------------------------------------

    model_results = model_results.sort_values(
        by=["Micro F1", "Macro F1", "Accuracy"],
        ascending=[False, False, False]
    ).reset_index(drop=True)

    # ------------------------------------------------------
    # Create Rank
    # ------------------------------------------------------

    model_results.insert(0, "Rank", range(1, len(model_results) + 1))

    # Move Rank column to first position
    cols = ["Rank"] + [
        c for c in model_results.columns
        if c != "Rank"
    ]

    model_results = model_results[cols]

    # ------------------------------------------------------
    # Round numeric values
    # ------------------------------------------------------

    numeric_cols = model_results.select_dtypes(
        include="number"
    ).columns

    model_results[numeric_cols] = (
        model_results[numeric_cols]
        .round(4)
    )

    # ------------------------------------------------------
    # Save Excel
    # ------------------------------------------------------

    model_results.to_csv(RESULT_FILE, index=False)

    print("=" * 60)
    print(f"{metrics['Model']} results saved successfully.")
    print(f"Results File : {RESULT_FILE}")
    print("=" * 60)

    return model_results
```

#**Evaluation Function**

```python
results = []

def evaluate_model(
    model_name,
    y_true,
    y_pred,
    y_prob=None
):

    metrics = {

        "Model":model_name,

        "Accuracy":
        accuracy_score(y_true,y_pred),

        "Micro Precision":
        precision_score(
            y_true,
            y_pred,
            average='micro',
            zero_division=0
        ),

        "Micro Recall":
        recall_score(
            y_true,
            y_pred,
            average='micro',
            zero_division=0
        ),

        "Micro F1":
        f1_score(
            y_true,
            y_pred,
            average='micro',
            zero_division=0
        ),

        "Macro F1":
        f1_score(
            y_true,
            y_pred,
            average='macro',
            zero_division=0
        ),

        "Weighted F1":
        f1_score(
            y_true,
            y_pred,
            average='weighted',
            zero_division=0
        ),

        "Hamming Loss":
        hamming_loss(
            y_true,
            y_pred
        ),

        "Jaccard Score":
        jaccard_score(
            y_true,
            y_pred,
            average='samples',
            zero_division=0
        )
    }

    if y_prob is not None:
        metrics["ROC-AUC"] = roc_auc_score(
            y_true,
            y_prob,
            average="micro"
        )
    else:
        metrics["ROC-AUC"] = np.nan

    results.append(metrics)

    update_results_to_csv(metrics)

    print("="*70)
    print(model_name)
    print("="*70)

    for k,v in metrics.items():

        if k!="Model":

            print(f"{k:<20}: {v:.4f}")
```
